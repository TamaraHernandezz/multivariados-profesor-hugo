Objetivo: implementar regresión lineal usando solo numpy

```python
#librerias necesarias
import numpy as np
```

```python
#1 generar dataset sintético con parámetros especficados
np.random.seed(42)
n=150 #número observaciones (datos)
x= np.random.rand(n,1)*10  #genera 150 random numbers entre 0 y 10
e= np.random.randn(n,1)*2  #genera el ruido e con desviación 2 normal
y= 2*x + 3 + e #regresión
```

```python
import matplotlib.pyplot as plt
```

```python
#checar e
print("media de e: ", np.mean(e))
print(e[:10])
print("desviacón estandar de e: ", np.std(e))
```

```python
#exploración de los datos
print("largo de x", x.shape)
print("largo de y", y.shape)
media_x= np.mean(x)
media_y= np.mean(y)
#estadísticas
print("media de x: ", media_x)
print("media de y: ", media_y)
print("desviacón estandar de x: ", np.std(x))
print("desviacón estandar de y: ", np.std(y))

#visualización en gráfica de distribución de los datos
plt.figure(figsize=(10,6))
plt.plot(x,y,"o", alpha=0.6, color='blue', label='Datos generados sintéticamente')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Distribución de los datos')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
#2. calcular varianza
varianza_x=np.var(x)
varianza_e=np.var(e)
varianza_y=np.var(y)
print("varianza x: ", varianza_x)
print("varianza e: ", varianza_e)
print("varianza y: ", varianza_y)
```

```python
#3. estimación de betas usando MCO
Sxx= np.sum((x-media_x)**2)
Sxy= np.sum((x-media_x)*(y-media_y)) #covarianza
print("Sxx:",Sxx)
pendiente_B1= (Sxy)/(Sxx)
intercepción_B0= media_y- pendiente_B1*media_x
print("pendiente B1: ", pendiente_B1) #real es 2
print("intercepción B0: ", intercepción_B0) #real es 3
```

```python
#4. calcular R^2 y MSE

#prediccion
y_pred=pendiente_B1*x+intercepción_B0
r2=(np.sum((y_pred-media_y)**2))/(np.sum((y-media_y)**2))
mse=np.mean((y-y_pred)**2)

print("R^2: ", r2)
print("MSE: ", mse)
```

```python
#5. visualización resultados
plt.figure(figsize=(10,6))
plt.plot(x,y,"o", alpha=0.6, color='blue', label='Datos generados sintéticamente')
plt.plot(x, y_pred, color='red', linewidth=2, label='Línea de regresión (MCO)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Visualización de resultados')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
